```python
import pandas as pd
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
"Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

revenue_values = [42, 45, 50, 48, 55, 60, 58, 62, 65, 70, 75, 85]
expense_values = [38, 40, 42, 47, 50, 52, 55, 60, 61, 64, 66, 72]
```

**Part A**

```python
#create revenue and expenses series
revenue = pd.Series(revenue_values, index=months, name="Revenue")
expenses = pd.Series(expense_values, index=months, name="Expenses")
print(revenue,"\n")
print(expenses)
```

```python
#Let a = [42, 45, 50] and b = [38, 40, 42]. Display a + b
a = [42, 45, 50]
b = [38, 40, 42]
print(a + b)
```

```python
#Calculate profit for all 12 months using a for loop that appends to a list.
profit=[]
for i in range(len(revenue_values)):
  profit.append(revenue_values[i]-expense_values[i])

print(profit)
```

```python
#Calculate Profit Using a Series Expression
prof = revenue - expenses
print(prof)
print(prof.tolist() == profit)
```

**Part B**

```python
#name prof to profit and print
prof.name="Profit"
print(prof)
```

```python
#create margin,the profit marginnn % as profit / revenue *100 ,rounded to 1 decimal
margin = (prof/revenue)*100
margin = margin.round(1)
print(margin)
```

```python
#calculate total revenue,total expense ,total profit and the overall margin % for the year
total_revenue = revenue.sum()
total_expenses = expenses.sum()
total_profit = profit.sum()
overall_margin = round((total_profit / total_revenue) * 100, 1)
print("Total Revenue:", total_revenue, "lakh")
print("Total Expenses:", total_expenses, "lakh")
print("Total Profit:", total_profit, "lakh")
print("Overall Profit Margin:", overall_margin, "%")
```

```python
#Find the best and worst months (name and profit) using a boolean condition on profit.max() and profit.min().
best_month = prof[prof == prof.max()]
worst_month = profit[profit == profit.min()]
print("Best Month:","\n",best_month)
print("\n")
print("Worst Month:","\n",worst_month)
```

```python
#Display the top 3 months by profit using sort_values().
print(profit.sort_values(ascending=False).head(3))
```

```python
#Calculate the average monthly profit and display the months that exceed it
avg_profit = profit.mean()
print("Average Monthly Profit:", round(avg_profit, 2), "lakh")
print("Months Above Average Profit:")
print(profit[profit > avg_profit])
```

**Part D**

```python
#Define profit_status(p): 8 or above = "Strong", 4 to below 8 = "Stable", below 4 = "Weak". Apply it to profit
def profit_status(p):
    if p >= 8:
        return "Strong"
    elif p >= 4:
        return "Stable"
    else:
        return "Weak"

status = profit.apply(profit_status)
print(status)
```

```python
#Display the number and percentage of months in each status.
status_count = status.value_counts()
status_percentage = (status.value_counts(normalize=True) * 100).round(1)
print("Number of Months:")
print(status_count)
print("\nPercentage of Months:")
print(status_percentage)
```

```python
#Define tax_payable(p): 25% of profit if profit is positive, otherwise 0. Apply it to profit
def tax_payable(p):
    if p > 0:
        return p * 0.25
    else:
        return 0

tax = profit.apply(tax_payable)
print(tax)
```

```python
#Calculate total tax and the post-tax profit for the year
total_tax = tax.sum()
posttax_profit = profit.sum() - total_tax
print("Total Tax:", total_tax, "lakh")
print("Post-Tax Profit:", posttax_profit, "lakh")
```
