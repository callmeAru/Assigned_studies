```python
import numpy as np

students = ["Anu", "Binu", "Cathy", "Deepa", "Ebin", "Fahad"]

marks = np.array([
    [82, 75, 91, 88],
    [61, 68, 70, 65],
    [45, 52, 48, 55],
    [90, 94, 89, 92],
    [72, 80, 76, 85],
    [38, 42, 49, 44]
])

subjects = ["Python", "SQL", "Excel", "Statistics"]
```

```python
#Total and average
total = np.sum(marks, axis=1)
avg = np.mean(marks, axis=1)
print("Total:", total)
print("Average:", avg)
```

```python
#Subject averages
subject_avg = np.mean(marks, axis=0)
print("Subject averages:", np.round(subject_avg, 2))
```

```python
#Highest mark in each subject
highest = np.max(marks, axis=0)
print("Highest marks:", highest)

print("\n")
#Pass/Fail
status = marks >= 50
print("Pass/Fail:",status)
```

```python
# 5. Students passed every subject
passed_all = np.all(marks >= 50, axis=1)
print("Passed all:",np.array(students)[passed_all])

print("\n")
# 6. Average below 60
print("Average below 60:",np.array(students)[avg < 60])
```

```python
#grade
marks=np.array([95,82,67,45,30])

conditions = [marks >=80,marks >=60,marks>=50]
choice=["A","B","C"]
grade=np.select(conditions,choice,default="fail")
print("Grade :",grade)
```
