```python
import pandas as pd
```

```python
enrol_list = [
    "Python", "SQL", "Excel", "Python", "Power BI", "SQL", "Python",
    "Excel", "Tableau", "Python", "SQL", "Python", "Power BI", "Excel",
    "Python", "SQL", "Statistics", "Python", "Excel", "SQL", "Power BI",
    "Python", "SQL", "Excel", "Python", "Tableau", "SQL", "Power BI",
    "Python", "Excel", "Python", "SQL", "Power BI", "Statistics", "Python",
    "Tableau", "SQL", "Excel", "Machine Learning", "Power BI"
]

mode_list = [
    "Online", "Classroom", "Online", "Online", "Online", "Online", "Hybrid",
    "Online", "Online", "Online", "Classroom", "Online", "Hybrid", "Online",
    "Classroom", "Online", "Hybrid", "Hybrid", "Hybrid", "Hybrid", "Online",
    "Hybrid", "Online", "Hybrid", "Classroom", "Online", "Online", "Online",
    "Classroom", "Online", "Hybrid", "Classroom", "Hybrid", "Online",
    "Classroom", "Online", "Online", "Hybrid", "Online", "Online"
]

rating_list = [
    5, 4, 4, 3, 5, 2, 4, 5, 3, 4, 1, 5, 4,
    3, 4, 5, 2, 4, 5, 4, 3, 4, 5, 2, 4
]
```

```python
enrole = pd.Series(enrol_list, name="Course Enrolled")
print(enrole)
```

```python
#display no: of record and assigned name
print("Total records:", len(enrole))
print("Series name:", enrole.name)
```

```python
#display the first five and last five record
print(f'First five records: { enrole.iloc[:5] }')
print(f'\nLast five records: { enrole.iloc[-5:] }')
```

```python
#display the no: of enrolments recieved by each course
course_count = enrole.value_counts()
print(course_count)
```

```python
#display each course % share
share = (enrole.value_counts(normalize=True)*100).round(1)
print(share)
```

```python
#display distinct course and their total no:
print("Distinct courses:",enrole.unique())
print("\n")
print("Total distinct courses:", enrole.nunique())
```

```python
#find the most popular course
print("Most popular course:", course_count.idxmax())
```

```python
#display the three most popular course
print(course_count.head(3))
```

```python
#display course whose share is below 10%
print(share[share < 10])
```

```python
#Prove that the two most popular courses account for more than half of all enrolments
top_two_share = share.head(2).sum()
print("Combined share:", top_two_share, "%")
print("More than half of enrolments:", top_two_share > 50)
```

```python
#create a series for study mode
mode = pd.Series(mode_list, name="Study Mode")
print(mode)
```

```python
#Display the number of learners and percentage share for each study mode
mode_count = mode.value_counts()
mode_share = (mode.value_counts(normalize=True) * 100).round(1)
print("Number of learners:","\n", mode_count)
print("\nPercentage share:","\n", mode_share)
```

```python
#Find the most preferred study mode and combined percentage
most_preferred = mode_count.idxmax()
print("Most preferred study mode:", most_preferred)
comb_share = mode_share.loc[["Online", "Hybrid"]].sum()
print("Online + Hybrid percentage:", comb_share)
```

```python
#Create Series and display statistics
ratings = pd.Series(rating_list, name="Ratings")
print("Number of ratings:", len(ratings))
print("Average rating:", round(ratings.mean(), 2))
print("Highest rating:", ratings.max())
print("Lowest rating:", ratings.min())
```

```python
#Count each rating
rating_count = ratings.value_counts().sort_index()
print(rating_count)
```

```python
#Function to classify sentiments
def sentiment(r):
    if r >= 4:
        return "Positive"
    elif r == 3:
        return "Neutral"
    else:
        return "Negative"

ratings_sentiment = ratings.apply(sentiment)
print(ratings_sentiment)
```

```python
#Percentage of each sentiment

sentiment_share = (ratings_sentiment.value_counts(normalize=True) * 100).round(1)
print(sentiment_share)
```

```python
# Function to decide portfolio action

def port_action(pct):
    if pct >= 20:
        return "Expand"
    elif pct >= 10:
        return "Maintain"
    else:
        return "Review"

action = share.apply(port_action)
print(action)
```

```python
portfolio = pd.DataFrame({
    "Share": share,
    "Action": action
})

print(portfolio)
```

```python
#Demonstrate inconsistent course names

new_courses = pd.Series(["python", " SQL"])
updated_courses = pd.concat([enrole, new_courses], ignore_index=True)

print("Frequency before cleaning:")
print(updated_courses.value_counts())
print("\nDistinct courses before cleaning:", updated_courses.nunique())
```

```python
#Clean course names

clean_courses = updated_courses.str.strip().str.title()
print("Frequency after cleaning:")
print(clean_courses.value_counts())
print("\nDistinct courses after cleaning:", clean_courses.nunique())
```
