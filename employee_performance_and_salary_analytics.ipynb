```python
import numpy as np

emp = ["Arun", "Bhavana", "Charan", "Divya", "Eshan", "Fathima"]

perf = np.array([
    [78, 82, 85, 88],
    [65, 70, 68, 74],
    [92, 89, 95, 91],
    [55, 62, 60, 67],
    [81, 79, 84, 87],
    [72, 75, 78, 80]
])

quarters = ["Q1", "Q2", "Q3", "Q4"]

print(emp)
```

```python
#total performance
total = np.sum(perf,axis=1)
print(total)
```

```python
#average quaterly performace for every employee
average=np.mean(perf,axis=1)
print(f'Average : ',average)
```

```python
#averageperformance for each quarter
print(f'Quater Average : {np.mean(perf,axis=0)}')
```

```python
#employee with highest total performance
cal = np.argmax(np.sum(perf,axis=1))
print(cal ,":", emp[cal])
```

```python
#rank all employee
rank=np.argsort(total[::-1])
for i in rank:
  print(emp[i],":",total[i])
```

```python
#percentage growth
growth = (perf[:, 1:] - perf[:, :-1]) / perf[:, :-1] * 100
print("Growth:",np.round(growth, 2))
```

```python
#avg equal or less than 80
print("Average equal or less than 80 :",np.array(emp)[average >= 80])
```

```python
#employee whose avg is below 80
print(np.sum(average<70))
```

```python
#highest performance
print(np.max(perf))
```

```python
#employee and quater where highest score
row,col= np.unravel_index(np.argmax(perf), perf.shape)

print("Employee:", emp[row])
print("Quarter:", quarters[col])
```
