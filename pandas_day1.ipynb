```python
import pandas as pd
```

```python
Customer=pd.Series({'C001':8,'C002':6,'C003':9,'C004':4,'C005':7,'C006':10,'C007':5,'C008':8})
print('All values :\n',Customer)
```

```python
print('Indexs of series :',Customer.index)
```

```python
print('Data type of series :',Customer.dtype)
```

```python
Customer.name="Customer Satisfaction"
print('Changed name of Customer series :',Customer.name)
```

```python
print('Satisfaction score of C003 :',Customer.loc["C003"])
```

```python
print('Satisfaction Score of 6th customer :',Customer.iloc[5])
```

```python
print('Scores of Customers :',Customer.loc[["C001","C003","C006","C008"]])
```

```python
print('Scores of position 0,2,5,7 :\n',Customer.iloc[[0,2,5,7]])
```

```python
print('Scores of customer C002 and C007 :\n',Customer.loc[["C002","C007"]])
```

```python
print('Total Number of Customer :',Customer.count())
```

```python
print('Total satisfaction Score : ',Customer.sum())
```

```python
print('Average Satisfaction Score :',Customer.mean())
```

```python
print("Highest Satisfaction Score :",Customer.max())
```

```python
print('Lowest Satisfaction score :',Customer.min())
```

```python
print('Median Satisfaction Score :',Customer.median())
```

```python
print('All customer whose satisfaction score is greater than or equal to 8 :\n',Customer[Customer>=8])
```

```python
print('All customer whose satisfaction score is less than 6 ;\n',Customer[Customer<=6])
```
