```python
#What is the size and structure of the dataset?
df = pd.read_csv(
    "/kaggle/input/datasets/vivek468/superstore-dataset-final/Sample - Superstore.csv",
    encoding="latin1"
)

df.head()
```

```python
#Are there missing values or duplicate records?
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nDataset Information:")
df.info()

print("\nFirst 5 Rows:")
print(df.head())
```

```python
#Are There Missing Values or Duplicate Records?
print("Missing Values:")
print(df.isnull().sum())

print("\nTotal Missing Values:")
print(df.isnull().sum().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())
```

```python
#What Are the Total Sales, Total Profit, and Average Order-Line Sales?
print("Total Sales:", df['Sales'].sum())
print("Total Profit:", df['Profit'].sum())
print("Average Order-Line Sales:", round(df['Sales'].mean(), 2))
```

```python
#Which product categories generate the most sales?
category_sales = df.groupby('Category')['Sales'].sum()
print(category_sales.sort_values(ascending=False))
```

```python
print("Top Sales Category:", category_sales.idxmax())
```

```python
#Which sub-categories generate the highest and lowest profit?
subcategory_profit = df.groupby('Sub-Category')['Profit'].sum()

print("Profit by Sub-Category:")
print(subcategory_profit.sort_values(ascending=False))
print("\nHighest Profit:")
print(subcategory_profit.idxmax(), subcategory_profit.max())

print("\nLowest Profit:")
print(subcategory_profit.idxmin(), subcategory_profit.min())
```

```python
#Which region has the highest sales?
region_sales = df.groupby('Region')['Sales'].sum()
print(region_sales.sort_values(ascending=False))
print("\nHighest Sales Region:", region_sales.idxmax())
```

```python
#Which customer segment contributes the most sales?
segment_sales = df.groupby('Segment')['Sales'].sum()
print(segment_sales.sort_values(ascending=False))
print("\nTop Customer Segment:", segment_sales.idxmax())
```

```python
#How do discounts relate to profit?
discount_profit = df.groupby('Discount')['Profit'].mean()
print(discount_profit.sort_index())
```

```python
#How do sales change over time?
df['Order Date'] = pd.to_datetime(df['Order Date'])
monthly_sales = df.groupby(
    df['Order Date'].dt.to_period('M')
)['Sales'].sum()

print(monthly_sales)
```
