```python
import numpy as np

stores = ["Store_A", "Store_B", "Store_C", "Store_D", "Store_E"]

sales = np.array([
    [120000, 135000, 142000, 155000],
    [98000, 105000, 110000, 108000],
    [150000, 148000, 160000, 172000],
    [87000, 92000, 99000, 115000],
    [130000, 138000, 141000, 149000]
])
```

```python
#calculate total sales for every store
total = np.sum(sales, axis=1)
print(total)
```

```python
#average quarterly sales for every store
avg = np.mean(sales, axis=1)
print(avg)
```

```python
#total sales for every quater
q_total = np.sum(sales, axis=0)
print(q_total)
```

```python
#stores with highest annula sales
index = np.argmax(total)
print(stores[index],":",total[index])
```

```python
#rank stores by annual sales
rank = np.argsort(total[::-1])

for i in rank:
    print(stores[i],":",total[i])
```

```python
#Quarter-over-quarter growth
growth = (sales[:, 1:] - sales[:, :-1]) / sales[:, :-1] * 100
print("Growth:",np.round(growth, 2))
print("\n")

#Above overall average
overall_average = np.mean(total)
print("Above average:",np.array(stores)[total > overall_average])
```
