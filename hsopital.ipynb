```python
hos = pd.read_csv('/content/Hospital_Patients.csv')
```

```python
#Load the file and take a first look at the data: first and last rows, a random sample, shape, columns and data types. List every problem you can spot (wrong data types, missing values, anything odd). Age is shown as 'object' - explain why.
print(hos.head(),"\n")
print(hos.tail(),"\n")
print("random :","\n",hos.sample(5,random_state=1),"\n")
print(hos.shape,"\n")
print(hos.columns,"\n")
print(hos.dtypes,"\n")
print(hos.isnull().sum())
```

```python
#Produce summary statistics for the numeric columns and for all columns. Report the mean, median, mode, minimum, maximum, standard deviation, count and sum of StayDays. What is the total number of patient-days?
print("summary for all numeric columns : ")
print(hos.describe())
print(hos.describe(include='all'))
print("Mean:", hos['StayDays'].mean())
print("Median:", hos['StayDays'].median())
print("Mode:", hos['StayDays'].mode()[0])
print("Minimum:", hos['StayDays'].min())
print("Maximum:", hos['StayDays'].max())
print("Standard Deviation:", hos['StayDays'].std(),"\n")
print("Count:", hos['StayDays'].count())
print("Sum:", hos['StayDays'].sum())
```

```python
# Compare the mean and median of BillAmount.
#Which one better represents a typical bill, and why?

print("Mean Bill Amount:", hos['BillAmount'].mean())
print("Median Bill Amount:", hos['BillAmount'].median())
```

```python
#How many patients are in each Department (count and percentage)?
#What percentage of patients have insurance?
#How many unique values does Gender have?

print("Department Count:")
print(hos['Department'].value_counts())

print("\nDepartment Percentage:")
print(hos['Department'].value_counts(normalize=True) * 100)

print("\nInsurance Percentage:")
print(hos['Insurance'].value_counts(normalize=True) * 100)

print("\nUnique Gender Values:")
print(hos['Gender'].nunique())
```

```python
#Create a missing-value report showing the count and percentage for each column.
#Decide how to treat BillAmount and SatisfactionScore, and fill both with the median.
#Justify why the mean or zero is not suitable.

print("Missing Values:")
print(hos.isnull().sum())

print("\nMissing Percentage:")
print((hos.isnull().sum() / len(hos)) * 100)

print("\nTotal Missing Values:")
print(hos.isnull().sum().sum())

hos['BillAmount'] = hos['BillAmount'].fillna(hos['BillAmount'].median())
hos['SatisfactionScore'] = hos['SatisfactionScore'].fillna(hos['SatisfactionScore'].median())

print("\nMissing Values After Treatment:")
print(hos.isnull().sum())
```

```python
#Detect, display and remove the duplicate patient records.
#Then check that PatientID is unique.

print("Duplicate Rows:", hos.duplicated().sum())
display(hos[hos.duplicated(keep=False)])
hos = hos.drop_duplicates()
print("Shape After Removing Duplicates:", hos.shape)
print("PatientID Unique:", hos['PatientID'].is_unique)
```

```python
# Convert Age to a number.Some ages are the text 'unknown' - decide how to handle them.
# After conversion, check the missing values again and explain why new missing values appeared.
# Fill them using the median age and convert Age to int.

hos['Age'] = pd.to_numeric(hos['Age'], errors='coerce')

print("Missing Age Values:",hos['Age'].isnull().sum())
hos['Age'] = hos['Age'].fillna(hos['Age'].median())
hos['Age'] = hos['Age'].astype(int)
print("Age Data Type:", hos['Age'].dtype)
```

```python
#Convert AdmitDate to datetime so that invalid dates become NaT.How many records have invalid dates?
#Also convert Department, Gender and Insurance to the 'string' type.

hos['AdmitDate'] = pd.to_datetime(hos['AdmitDate'], errors='coerce')
print("Invalid Dates:", hos['AdmitDate'].isnull().sum())
hos['Department'] = hos['Department'].astype('string')
hos['Gender'] = hos['Gender'].astype('string')
hos['Insurance'] = hos['Insurance'].astype('string')

print("\nUpdated Data Types:")
print(hos.dtypes)
```

```python
#Which patients stayed more than 10 days? Which patients are unhappy (SatisfactionScore below 5)?
# Which uninsured patients have a bill above Rs. 50,000?

print("Patients Who Stayed More Than 10 Days:")
display(hos[hos['StayDays'] > 10])
print("Patients with Satisfaction Below 5:")
display(hos[hos['SatisfactionScore'] < 5])
print("Uninsured Patients with Bill Above 50000:")
display(hos[(hos['Insurance'] == 'No') &(hos['BillAmount'] > 50000)])
```

```python
#Show the 5 most recently admitted patients and the 5 patients.with the highest BillAmount. Does the highest bill look suspicious?

print("5 Most Recently Admitted Patients:")
display(hos.sort_values('AdmitDate', ascending=False).head(5))
print("5 Patients with Highest Bill Amount:")
display(hos.nlargest(5, 'BillAmount'))
```

```python
# Task 11: For each Department, find the average StayDays and average BillAmount.
# For each Department, also give the count, mean, median, minimum and maximum of BillAmount.
# For Insurance Yes/No, find the average SatisfactionScore.
# Which department is the most expensive?

print("Average Stay Days by Department:")
print(hos.groupby('Department')['StayDays'].mean())
print("\nAverage Bill Amount by Department:")
print(hos.groupby('Department')['BillAmount'].mean())
print("\nBill Amount Summary:")
display(hos.groupby('Department')['BillAmount'].agg(['count', 'mean', 'median', 'min', 'max']))
print("\nAverage Satisfaction by Insurance:")
print(hos.groupby('Insurance')['SatisfactionScore'].mean())
print("\nMost Expensive Department:")
print(hos.groupby('Department')['BillAmount'].mean().idxmax())
```

```python
#Find the correlation between StayDays and BillAmount,
#between StayDays and SatisfactionScore, and between Age and StayDays.
#Explain what each value means for the hospital.

print("StayDays and BillAmount Correlation:",hos['StayDays'].corr(hos['BillAmount']))
print("\nStayDays and SatisfactionScore Correlation:",hos['StayDays'].corr(hos['SatisfactionScore']))
print("\nAge and StayDays Correlation:",hos['Age'].corr(hos['StayDays']))
```

```python
#Detect outliers in BillAmount using the IQR method.Show the limits and the outlier rows.
# Then compute the correlation between StayDays and BillAmount again after removing the outlier. What changed? Which version would you trust?

Q1 = hos['BillAmount'].quantile(0.25)
Q3 = hos['BillAmount'].quantile(0.75)

IQR = Q3 - Q1
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)

outliers = hos[(hos['BillAmount'] < lower_limit) |(hos['BillAmount'] > upper_limit)]

print("\nOutlier Rows:")
print(outliers)

print("Correlation Before Removing Outliers:",hos['StayDays'].corr(hos['BillAmount']))
hos_clean = hos[(hos['BillAmount'] >= lower_limit) &(hos['BillAmount'] <= upper_limit)].copy()

print("\nCorrelation After Removing Outliers:",hos_clean['StayDays'].corr(hos_clean['BillAmount']))
print("\nOriginal Shape:", hos.shape)
print("Cleaned Shape:", hos_clean.shape)
```

```python
#Create Cost_Per_Day as BillAmount divided by StayDays.Which Department has the highest average Cost_Per_Day?

hos_clean['Cost_Per_Day'] = (hos_clean['BillAmount'] / hos_clean['StayDays'])
print("Average Cost Per Day by Department:")
print(hos_clean.groupby('Department')['Cost_Per_Day'].mean())
print("\nDepartment with Highest Average Cost Per Day:",hos_clean.groupby('Department')['Cost_Per_Day'].mean().idxmax())
```

```python
#Using lambda functions, create Stay_Type:'Short' for 3 days or fewer, 'Medium' for 4-7 days, and 'Long' for more than 7 days.
# Create Age_Group: 'Child' (below 18), 'Adult' (18-59) and 'Senior' (60 and above). Count each category.

hos_clean['Stay_Type'] = hos_clean['StayDays'].apply(lambda x: 'Short' if x <= 3 else 'Medium' if x <= 7 else 'Long')
hos_clean['Age_Group'] = hos_clean['Age'].apply(lambda x: 'Child' if x < 18 else 'Adult' if x < 60 else 'Senior')

print("Stay Type Count:")
print(hos_clean['Stay_Type'].value_counts())
print("\nAge Group Count:")
print(hos_clean['Age_Group'].value_counts())
```

```python
# Task 16: Find the average SatisfactionScore for each Stay_Type and each Age_Group. What pattern do you see?

print("Average Satisfaction by Stay Type:")
print(hos_clean.groupby('Stay_Type')['SatisfactionScore'].mean())
print("\nAverage Satisfaction by Age Group:")
print(hos_clean.groupby('Age_Group')['SatisfactionScore'].mean())
```

```python
#Export hos_clean to cleaned_hospital.csv without the index,then reload the file to verify its shape and data types.

hos_clean.to_csv('cleaned_hospital.csv', index=False)
new_df = pd.read_csv('cleaned_hospital.csv')
print("Reloaded DataFrame Shape:")
print(new_df.shape)
print("\nData Types:")
print(new_df.dtypes)
```

```python
# Final Question 1: Which department needs attention - the one with the highest cost per day, or the one with the lowest satisfaction?

cost_dept = hos_clean.groupby('Department')['Cost_Per_Day'].mean()
satisfaction_dept = hos_clean.groupby('Department')['SatisfactionScore'].mean()

print("Average Cost Per Day:")
print(cost_dept)
print("\nDepartment with Highest Cost Per Day:")
print(cost_dept.idxmax())
print("\nAverage Satisfaction:")
print(satisfaction_dept)
print("\nDepartment with Lowest Satisfaction:")
print(satisfaction_dept.idxmin())
```

```python
# Final Question 2: Does a longer stay reduce satisfaction? Support your answer with numbers.

corr = hos_clean['StayDays'].corr(hos_clean['SatisfactionScore'])
print("Correlation:", corr)
print("Average Satisfaction by Stay Type:")
print(hos_clean.groupby('Stay_Type')['SatisfactionScore'].mean())
```
