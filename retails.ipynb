```python
#Use the Interquartile Range (IQR) method to identify outliers in Quantity. Print Q1, Q3, the lower and upper limits, and the outlier rows.
correlation = df[['Quantity', 'Discount', 'UnitPrice', 'Rating']].corr()
print("Correlation Matrix:")
display(correlation)
print("Quantity and Discount Correlation:",correlation.loc['Quantity', 'Discount'])
print("UnitPrice and Rating Correlation:",correlation.loc['UnitPrice', 'Rating'])
```

```python
#Examine the outlier rows and decide whether they are genuine or data-entry errors, giving your reasoning. Create a DataFrame named df_clean that excludes the rows you judge to be errors, and compare the mean of Quantity before and after.
Q1 = df['Quantity'].quantile(0.25)
Q3 = df['Quantity'].quantile(0.75)
IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
outliers = df[(df['Quantity'] < lower_limit) |(df['Quantity'] > upper_limit)]
display(outliers)
```

```python
#Examine the outlier rows and decide whether they are genuine or data-entry errors, giving your reasoning. Create a DataFrame named df_clean that excludes the rows you judge to be errors, and compare the mean of Quantity before and after.
df_clean = df[
    (df['Quantity'] >= lower_limit) &
    (df['Quantity'] <= upper_limit)].copy()

print("Mean Before Removing Outliers:", df['Quantity'].mean())
print("Mean After Removing Outliers:", df_clean['Quantity'].mean())
print("Shape Before:", df.shape)
print("Shape After:", df_clean.shape)
```

```python
# Use df_clean from this point onwards. Create a column Gross (Quantity x UnitPrice) and a column Revenue, which is Gross after applying the Discount percentage.
df_clean['Gross'] = df_clean['Quantity'] * df_clean['UnitPrice']
df_clean['Revenue'] = (df_clean['Gross'] * (1 - df_clean['Discount'] / 100))
display(df_clean[['OrderID', 'Gross', 'Revenue']])
```

```python
#Create a column Order_Size classifying each order as 'Large' if Revenue is at least 10000, 'Medium' if Revenue is at least 2000 but below 10000, and 'Small' otherwise. Count the orders in each size class.
df_clean['Order_Size'] = df_clean['Revenue'].apply(
    lambda x: 'Large' if x >= 10000
    else 'Medium' if x >= 2000
    else 'Small')

print("Orders in Each Size Class:")
print(df_clean['Order_Size'].value_counts())
```

```python
#  Find the top 5 orders by Revenue and the total Revenue for each Category.
print("Top 5 Orders by Revenue:")
display(df_clean.nlargest(5, 'Revenue'))

print("Total Revenue for Each Category:")
category_revenue = df_clean.groupby('Category')['Revenue'].sum()
print(category_revenue)
```
