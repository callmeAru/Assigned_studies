a college keeps student details exam marks and attendance in 3 separate csv file all are linked by student id

q1: load all 3 csv file

q2 : combine the 3 datasets using student id

q3: cal total marks

q4: cal avg marks

q5: identify students with an avg mark of 80

q6: identify student whose attendance is below 75%

q7: save the final dataset as student_performance.csv

```python
from google.colab import files
uploaded=files.upload()
```

```python
stud=pd.read_csv('/content/student_details.csv')
exm=pd.read_csv('/content/exam_marks.csv')
attend=pd.read_csv('/content/attendance.csv')
```

```python
print(stud,"\n\n",exm,"\n\n",attend)
```

```python
merged = stud.merge(exm, on="Student_ID").merge(attend, on="Student_ID")
print(merged)
```

```python
merged['Total_mark']=merged['Python']+merged['SQL']
print(merged['Total_mark'])
```

```python
merged['Avg_mark']=merged['Total_mark']/2
print(merged['Avg_mark'])
```

```python
top = merged[merged['Avg_mark'] >= 80]
print("Students with Avg Marks >= 80:")
print(top[['Student_ID', 'Name', 'Avg_mark']])
```

```python
low=merged[merged['Attendance_Percentage']<75]
print("Student with low attendance percentage : ")
print(low[['Student_ID','Name']])
```

```python
merged.to_csv('student_performance.csv')
```

a supermarket daly save 1 csv per month with identical columns the manager wants a single quarterly report

q1: load the 3 monthly files

q2: stack them into on dataset and record which month each row came from

q3: cal revenue (unit sol*unit price)

q4: find the total reveue per product for the quarter

q5; find the month with highest total revenue

q6: save as quarterly_sales.csv

```python
jan=pd.read_csv('/content/sales_jan.csv')
feb=pd.read_csv('/content/sales_feb.csv')
mar=pd.read_csv('/content/sales_mar.csv')
```

```python
print(jan,"\n\n",feb,"\n\n",mar)
```

```python
quarter = pd.concat([jan,feb,mar],keys=["jan","feb","mar"])
print(quarter)
```

```python
quarter['Revenue']=quarter['Unit_Price']*quarter['Units_Sold']
print(quarter[['Product','Revenue']])
```

```python
print(quarter)
```

```python
quarter=quarter.reset_index(level=0).rename(columns={"level_0":"Month"})
print(quarter)
```

```python
total_jan=(jan['Units_Sold']*jan['Unit_Price']).sum()
print("total revenue :",total_jan)
total_feb=(feb['Units_Sold']*feb['Unit_Price']).sum()
print("Total revenue :",total_feb)
total_mar=(mar['Units_Sold']*mar['Unit_Price']).sum()
print("Total revenue :",total_mar)
```

```python
highest = max(total_jan,total_feb,total_mar)
print(highest)
```

```python
value = quarter.groupby('Month').max()
print(value)
```

```python
m_revenue = quarter.groupby('Month')['Revenue'].sum()
highest_month = m_revenue.idxmax()
print("Total Revenue per Month:","\n",m_revenue)
print(f"\nMonth with the highest total revenue: {highest_month}")
```

```python
quarter.to_csv('quarterly.csv')
```
