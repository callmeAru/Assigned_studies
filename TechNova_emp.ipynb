```python
#Show the rows where Salary is missing. Then fill the missing Salary and missing PerformanceScore values with their respective median values. Explain why the median is safer than the mean for Salary.
display(Hr[Hr['Salary'].isnull()])
Hr['Salary'] = Hr['Salary'].fillna(Hr['Salary'].median())
Hr['PerformanceScore'] = Hr['PerformanceScore'].fillna(Hr['PerformanceScore'].median())
print(Hr.isnull().sum())
```

```python
#Find the correlation of Experience with Salary, Attendance with PerformanceScore, and Age with Experience. Interpret each value (strong / weak, positive / negative).
print("Experience and Salary:",Hr['Experience'].corr(Hr['Salary']))
print("Attendance and Performance:",Hr['Attendance'].corr(Hr['PerformanceScore']))
print("Age and Experience:",Hr['Age'].corr(Hr['Experience']))
```

```python
#Use the IQR method to find the salary outliers. Show the lower and upper limits and the outlier rows. Do you think the outlier is a real salary or a typing error? How can you decide?
Q1 = Hr['Salary'].quantile(0.25)
Q3 = Hr['Salary'].quantile(0.75)
IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)

outliers = Hr[(Hr['Salary'] < lower_limit) | (Hr['Salary'] > upper_limit)]

display(outliers)
```

```python
#Remove the outlier and store the result in df_clean. Compute the Experience-Salary correlation again. How much did one extreme value change the result?
df_clean = Hr[(Hr['Salary'] >= lower_limit) & (Hr['Salary'] <= upper_limit)].copy()

print("Correlation Before:", Hr['Experience'].corr(Hr['Salary']))
print("Correlation After:", df_clean['Experience'].corr(df_clean['Salary']))
```

```python
#Create a column Overall_Score as the row-wise mean of PerformanceScore and Attendance, and a column Annual_Bonus as 10% of Salary.
df_clean['Overall_Score'] = df_clean[['PerformanceScore', 'Attendance']].mean(axis=1)
df_clean['Annual_Bonus'] = df_clean['Salary'] * 0.10
display(df_clean)
```

```python
#Using a lambda function, create a column Grade: 'Excellent' if PerformanceScore >= 85, 'Good' if PerformanceScore >= 70, otherwise 'Needs Improvement'. Show the count of employees in each grade and the average Salary for each grade.
df_clean['Grade'] = df_clean['PerformanceScore'].apply(
    lambda x: 'Excellent' if x >= 85
    else 'Good' if x >= 70
    else 'Needs Improvement')

print(df_clean['Grade'].value_counts())
print(df_clean.groupby('Grade')['Salary'].mean())
```

```python
#Export the cleaned DataFrame to cleaned_hr.csv without the index, and verify the export by reloading the file.
df_clean.to_csv('cleaned_hr.csv', index=False)
new_df = pd.read_csv('cleaned_hr.csv')
print(new_df.shape)
print(new_df.dtypes)
```
